```python
# load Cellosaurus
cellosaurus = pd.read_csv("../../data/nomenclature/7_cellosaurus.csv")
print(cellosaurus.shape)
print(cellosaurus.columns.tolist())
cellosaurus.head()
```

```python
# does Cellosaurus contain DepMap cross-references?
has_depmap = cellosaurus["Cross-references"].str.contains("DepMap", na=False)
print("Rows with a DepMap cross-reference:", has_depmap.sum())
print(cellosaurus.loc[has_depmap, "Cross-references"].head(3).tolist())
```

```python
import re

# pull the ACH id that follows "DepMap;" in the cross-references text
def extract_ach(xref):
    if not isinstance(xref, str):
        return None
    m = re.search(r"DepMap;\s*(ACH-\d+)", xref)
    return m.group(1) if m else None

cellosaurus["ach_from_xref"] = cellosaurus["Cross-references"].apply(extract_ach)

# how many rows now have an ACH id?
print("Cellosaurus rows with an ACH id:", cellosaurus["ach_from_xref"].notna().sum())
```

```python
orphans = sig_ids - ref_ids                      # the 251
cello_achs = set(cellosaurus["ach_from_xref"].dropna())

recovered = orphans & cello_achs
print(f"Orphans recoverable via Cellosaurus: {len(recovered)} of {len(orphans)}")
```

```python
recovered_rows = cellosaurus[cellosaurus["ach_from_xref"].isin(recovered)]
recovered_rows[["Identifier (cell line name)", "ach_from_xref", "Diseases", "Species of origin"]].head(10)
```

```python
recovered_meta = cellosaurus[cellosaurus["ach_from_xref"].isin(recovered)][
    ["ach_from_xref", "Identifier (cell line name)", "Diseases"]
].rename(columns={"ach_from_xref": "DepMap_ID"})
print(recovered_meta.shape)
recovered_meta.head(10)
```

```python
# build the table of recovered orphan cell lines
recovered_meta = cellosaurus[cellosaurus["ach_from_xref"].isin(recovered)][
    ["ach_from_xref", "Identifier (cell line name)", "Diseases", "Species of origin"]
].copy()

# rename to match the dim_cell_lines convention (ACH as the key)
recovered_meta = recovered_meta.rename(columns={
    "ach_from_xref": "DepMap_ID",
    "Identifier (cell line name)": "cell_line_name",
    "Diseases": "disease",
    "Species of origin": "species"
})

recovered_meta = recovered_meta.reset_index(drop=True)
print("Recovered orphan cell lines:", recovered_meta.shape)
recovered_meta
```

```python
print("Unique ACH IDs:", recovered_meta["DepMap_ID"].nunique(), "of", len(recovered_meta))
```

```python
recovered_meta.to_csv("../../processed/recovered_orphans_47.csv", index=False)
print("Saved 47 recovered orphans for dim_cell_lines enrichment")
```

```python
# HPA's unique cell line names
hpa_names = set(hpa["Cell line"].unique())

# sample_info's name columns: try matching against stripped names
# normalise HPA names the same way (remove punctuation/spaces, uppercase)
def strip_name(n):
    return str(n).upper().replace("-", "").replace(" ", "").replace("/", "").replace(".", "")

hpa_stripped = {strip_name(n) for n in hpa_names}
ref_stripped = set(sample_info["stripped_cell_line_name"].dropna().str.upper())

matched = hpa_stripped & ref_stripped
print(f"HPA cell lines matching sample_info (stripped): {len(matched)} of {len(hpa_stripped)}")
```
