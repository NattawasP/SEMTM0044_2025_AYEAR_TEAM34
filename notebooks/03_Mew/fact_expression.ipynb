### Fact_proteomics

```python
import pandas as pd

folder = r"C:\Users\Pattanun\Downloads\AZ_project_data\data\gene expression"
df_4 = pd.read_csv(folder + r"\4_Harmonized_MS_CCLE_Gygi_subsetted.csv")
print(df_4.shape)
print(df_4.head())
print(df_4.columns[:10].tolist())
```

```python
df_4 = df_4.rename(columns={"Unnamed: 0": "ACH_ID"})
```

```python
df_4_long = df_4.melt(
    id_vars=["ACH_ID"],
    var_name="protein_col",
    value_name="protein_intensity"
)
print(df_4_long.shape)
print(df_4_long.head())
```

```python
df_4_long["uniprot_id"] = df_4_long["protein_col"].str.extract(r"^([A-Za-z0-9\-]+)")
df_4_long["hugo_symbol"] = df_4_long["protein_col"].str.extract(r"\((.+)\)")

df_4_long = df_4_long.drop(columns=["protein_col"])
print(df_4_long.head())
```

```python
import re

folder = r"C:\Users\Pattanun\Downloads\AZ_project_data\data\gene expression"
df_2 = pd.read_csv(folder + r"\2_DepMap_OmicsExpressionAllGenesTPMLogp1Profile.csv")

gene_cols = df_2.columns[1:]  

hugo_to_ensembl = {}
for col in gene_cols:
    match = re.match(r"^(.+) \((ENSG\d+)\)$", col)
    if match:
        hugo_to_ensembl[match.group(1)] = match.group(2)

print(f"mapping all: {len(hugo_to_ensembl)}")
print(list(hugo_to_ensembl.items())[:5])
```

```python
df_4_long["ensembl_id"] = df_4_long["hugo_symbol"].map(hugo_to_ensembl)

matched = df_4_long["ensembl_id"].notna().sum()
total = len(df_4_long)
print(f"match #: {matched} / {total} ({matched/total*100:.1f}%)")
print(df_4_long.head())
```

```python
df_4_long["is_detected"] = df_4_long["protein_intensity"].notna() & (df_4_long["protein_intensity"] != 0)

print(df_4_long["is_detected"].value_counts())
```

```python
df_4_long = df_4_long[["ACH_ID", "ensembl_id", "uniprot_id", "hugo_symbol", "protein_intensity", "is_detected"]]
print(df_4_long.head())
print(df_4_long.shape)
```

```python
df_4_long["z_score"] = df_4_long.groupby("ensembl_id")["protein_intensity"].transform(
    lambda x: (x - x.mean()) / x.std()
)

print(df_4_long.head(10))
```

```python
df_4_long.head(100)
```

```python
nom_folder = r"C:\Users\Pattanun\Downloads\AZ_project_data\data\nomenclature"
df_9 = pd.read_csv(nom_folder + r"\9_DepMap_sample_info.csv")

ach_to_cvcl = dict(zip(df_9["DepMap_ID"], df_9["RRID"]))

df_4_long["CVCL_ID"] = df_4_long["ACH_ID"].map(ach_to_cvcl)

# check match %
matched = df_4_long["CVCL_ID"].notna().sum()
total = len(df_4_long)
print(f"match #: {matched} / {total} ({matched/total*100:.1f}%)")

# manage column order (new)
df_4_long = df_4_long[["ACH_ID", "CVCL_ID", "ensembl_id", "uniprot_id", "hugo_symbol", "protein_intensity", "is_detected", "z_score"]]
print(df_4_long.head())
```

```python
import pandas as pd

df_2 = pd.read_csv(folder + r"\2_DepMap_OmicsExpressionAllGenesTPMLogp1Profile.csv")
df_8 = pd.read_csv(nom_folder + r"\8_DepMap_OmicsProfiles.csv")

print(df_2.shape)
print(df_2.iloc[:5, :5])
```

```python
df_2 = df_2.rename(columns={"Unnamed: 0": "PR_ID"})

# filter only rna profile
df8_rna = df_8[df_8["Datatype"] == "rna"]
pr_to_ach = dict(zip(df8_rna["ProfileID"], df8_rna["ModelID"]))

df_2["ACH_ID"] = df_2["PR_ID"].map(pr_to_ach)

matched = df_2["ACH_ID"].notna().sum()
total = len(df_2)
print(f"match ACH: {matched} / {total} ({matched/total*100:.1f}%)")
print(df_2[["PR_ID", "ACH_ID"]].head())
```

```python
import re

# parse column names first 
gene_cols = df_2.columns[2:]  # ข้าม PR_ID, ACH_ID

hugo_list = []
ensembl_list = []
valid_cols = []

for col in gene_cols:
    match = re.match(r"^(.+) \((ENSG\d+)\)$", col)
    if match:
        valid_cols.append(col)
        hugo_list.append(match.group(1))
        ensembl_list.append(match.group(2))

print(f"parse # {len(valid_cols)} genes from all {len(gene_cols)}")
```

```python
# only columns that can parse 
df_2_clean = df_2[["PR_ID", "ACH_ID"] + valid_cols]

# melt 
df_2_long = df_2_clean.melt(
    id_vars=["PR_ID", "ACH_ID"],
    var_name="gene_col",
    value_name="log2_tpm_plus1"
)

print(df_2_long.shape)
print(df_2_long.head())
```

```python
# create dict from gene_col from hugo_symbol, ensembl_id
col_to_hugo = dict(zip(valid_cols, hugo_list))
col_to_ensembl = dict(zip(valid_cols, ensembl_list))

df_2_long["hugo_symbol"] = df_2_long["gene_col"].map(col_to_hugo)
df_2_long["ensembl_id"] = df_2_long["gene_col"].map(col_to_ensembl)

df_2_long = df_2_long.drop(columns=["gene_col"])

print(df_2_long.head())
print(df_2_long.shape)
```

```python
ach_to_cvcl = dict(zip(df_9["DepMap_ID"], df_9["RRID"]))

df_2_long["CVCL_ID"] = df_2_long["ACH_ID"].map(ach_to_cvcl)

# add columns
df_2_long["source"] = "depmap"
df_2_long["tpm"] = (2 ** df_2_long["log2_tpm_plus1"]) - 1
df_2_long["is_expressed"] = df_2_long["log2_tpm_plus1"] > 1

print(df_2_long.head())
```

```python
df_2_final = df_2_long[["ACH_ID", "CVCL_ID", "ensembl_id", "hugo_symbol", "PR_ID", "source", "tpm", "log2_tpm_plus1", "is_expressed"]]

print(df_2_final.head())
print(df_2_final.shape)
```

```python
df_3 = pd.read_csv(folder + r"\3_GEOexpression.txt", sep="\t")
df_10 = pd.read_csv(nom_folder + r"\10_GEOInfo.txt", sep="\t")

print(df_3.shape)
print(df_3.iloc[:5, :5])
```

```python
df_3_long = df_3.melt(
    id_vars=["Gene"],
    var_name="GSM_ID",
    value_name="tpm"
).rename(columns={"Gene": "ensembl_id"})

print(df_3_long.shape)
print(df_3_long.head())
```

```python
gsm_to_cvcl = dict(zip(df_10["Geo_accession"], df_10["Cellosaurus_ID"]))
cvcl_to_ach = dict(zip(df_9["RRID"], df_9["DepMap_ID"]))

df_3_long["CVCL_ID"] = df_3_long["GSM_ID"].map(gsm_to_cvcl)
df_3_long["ACH_ID"] = df_3_long["CVCL_ID"].map(cvcl_to_ach)

print(df_3_long.head())
```

```python
df_3_long["source"] = "geo"
df_3_long["log2_tpm_plus1"] = np.log2(df_3_long["tpm"] + 1)
df_3_long["is_expressed"] = df_3_long["log2_tpm_plus1"] > 1

print(df_3_long.head())
print(df_3_long.shape)
```

```python
df_3_final = df_3_long[["ACH_ID", "CVCL_ID", "ensembl_id", "GSM_ID", "source", "tpm", "log2_tpm_plus1", "is_expressed"]]
print(df_3_final.head())
print(df_3_final.shape)
```
